```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from kmeans import KMeans
```

```python
# 1. Generate Data
X, y = make_blobs(n_samples=500, centers=4, n_features=2, cluster_std=0.60, random_state=0)
```

```python
# 2. Apply K-Means
km = KMeans(K=4, max_iters=150, plot_steps=False)
y_pred = km.fit(X)
```

```python
# 3. Plot
plt.figure(figsize=(10, 7))

# Plot clusters
plt.scatter(X[:, 0], X[:, 1], c=y_pred, cmap='viridis', s=50, alpha=0.6, edgecolors='w', label='Data Points')

# Plot Centroids
centroids = np.array(km.centroids)
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', s=200, marker='X', label='Centroids')

plt.title("K-Means Clustering (Own Implementation)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
