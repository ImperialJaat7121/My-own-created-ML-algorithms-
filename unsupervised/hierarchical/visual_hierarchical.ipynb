```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons, make_blobs
from hierarchical import HierarchicalClustering
```

```python
# 1. Generate "Moons" Data (Hard for K-Means, Easy for Single Linkage)
X, y = make_moons(n_samples=150, noise=0.05, random_state=42)
```

```python
# 2. Run Hierarchical Clustering
# Single Linkage acts like a "chain", connecting neighbors.
hc = HierarchicalClustering(n_clusters=2, linkage='single')
y_pred = hc.fit_predict(X)
```

```python
# 3. Plot
plt.figure(figsize=(10, 6))
plt.scatter(X[:, 0], X[:, 1], c=y_pred, cmap='coolwarm', s=50, edgecolors='k')
plt.title("Hierarchical Clustering (Single Linkage) on Non-Linear Data")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
# 4. Dendrogram Concept (Optional / Experimental)
# Since we didn't build a full dendrogram plotter (complex),
# we visualize the cluster result directly.
print(f"Final Cluster Counts: {np.bincount(y_pred)}")
```
