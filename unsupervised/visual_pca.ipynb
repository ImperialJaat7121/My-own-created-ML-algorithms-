```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
from pca import PCA
```

```python
# 1. Load Data
data = datasets.load_iris()
X = data.data
y = data.target
target_names = data.target_names
```

```python
# 2. Apply PCA (4D -> 2D)
pca = PCA(n_components=2)
pca.fit(X)
X_projected = pca.transform(X)

x1 = X_projected[:, 0]
x2 = X_projected[:, 1]
```

```python
# 3. Plot
plt.figure(figsize=(10, 7))

# Scatter plot for each class
for c, i, target_name in zip("rgb", [0, 1, 2], target_names):
    plt.scatter(x1[y == i], x2[y == i], c=c, label=target_name, alpha=0.7, edgecolors='k')

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA from Scratch: Iris Dataset (4D -> 2D)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
# 4. Explained Variance Bar Chart
variance = pca.explained_variance_ratio()
labels = ['PC1', 'PC2']

plt.figure(figsize=(6, 4))
plt.bar(labels, variance, color=['blue', 'orange'])
plt.title("Explained Variance Ratio")
plt.ylabel("Percentage of Variance")
plt.ylim(0, 1)
for i, v in enumerate(variance):
    plt.text(i, v + 0.02, f"{v:.2f}", ha='center')
plt.show()
```
