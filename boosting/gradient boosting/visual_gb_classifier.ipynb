```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
from gradient_boosting_classifier import GradientBoostingClassifier
```

```python
# 1. Generate Data
X, y = make_circles(n_samples=200, noise=0.1, factor=0.4, random_state=42)
```

```python
# 2. Train Model
clf = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, max_depth=2)
clf.fit(X, y)
```

```python
# 3. Create Meshgrid
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.05),np.arange(y_min, y_max, 0.05))
```

```python
# 4. Predict
Z = clf.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)
```

```python
# 5. Plot
plt.figure(figsize=(8, 6))
plt.contourf(xx, yy, Z, alpha=0.4, cmap='viridis')
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis', edgecolors='k')
plt.title("Gradient Boosting Classifier (Own Implementation)\nfitting Concentric Circles")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
