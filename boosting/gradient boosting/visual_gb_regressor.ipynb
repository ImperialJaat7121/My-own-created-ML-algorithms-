```python
import numpy as np
import matplotlib.pyplot as plt
from gradient_boosting_regressor import GradientBoostingRegressor
```

```python
# 1. Generate Noisy Sin Wave
rng = np.random.RandomState(1)
X = np.linspace(0, 10, 100)[:, np.newaxis]
y = np.sin(X).ravel() + np.sin(3 * X).ravel() + rng.normal(0, 0.1, X.shape[0])
```

```python
# 2. Train Models with different Estimators to see evolution
reg_1 = GradientBoostingRegressor(n_estimators=1, learning_rate=1.0, max_depth=2)
reg_10 = GradientBoostingRegressor(n_estimators=10, learning_rate=0.5, max_depth=2)
reg_100 = GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, max_depth=2)

reg_1.fit(X, y)
reg_10.fit(X, y)
reg_100.fit(X, y)
```

```python
# 3. Predict
y_1 = reg_1.predict(X)
y_10 = reg_10.predict(X)
y_100 = reg_100.predict(X)
```

```python
# 4. Plot
plt.figure(figsize=(12, 6))
plt.scatter(X, y, c="k", label="Data", s=10, alpha=0.5)
plt.plot(X, y_1, c="red", label="1 Tree (Underfit)", linewidth=2, linestyle='--')
plt.plot(X, y_10, c="orange", label="10 Trees", linewidth=2)
plt.plot(X, y_100, c="green", label="100 Trees (Good Fit)", linewidth=2)
plt.title("Gradient Boosting Regression: Evolution of Fit")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```
