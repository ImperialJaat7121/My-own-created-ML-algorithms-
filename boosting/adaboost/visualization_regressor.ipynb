```python
import numpy as np
import matplotlib.pyplot as plt
from adaboost_regressor import AdaBoostRegressor
```

```python
# Generate Sin Wave Data with Noise
rng = np.random.RandomState(1)
X = np.linspace(0, 6, 100)[:, np.newaxis]
y = np.sin(X).ravel() + np.sin(6 * X).ravel() + rng.normal(0, 0.1, X.shape[0])
```

```python
# Train
regr = AdaBoostRegressor(n_estimators=50)
regr.fit(X, y)
y_pred = regr.predict(X)
```

```python
# Plot
plt.figure(figsize=(10, 6))
plt.scatter(X, y, c="k", label="Training samples", s=15, alpha=0.6)
plt.plot(X, y_pred, c="red", label="AdaBoost Prediction", linewidth=2)
plt.xlabel("Data")
plt.ylabel("Target")
plt.title("AdaBoost Regressor (Own Implementation) fitting Sin Wave")
plt.legend()
plt.show()
```
