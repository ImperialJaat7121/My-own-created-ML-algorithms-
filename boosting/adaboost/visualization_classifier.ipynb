```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from adaboost_classifier import AdaBoostClassifier
```

```python
# 1. Generate Non-Linear Data
X, y = make_moons(n_samples=200, noise=0.20, random_state=42)
# Convert 0 -> -1 for the logic
y = np.where(y == 0, -1, 1)
```

```python
# 2. Train Model
clf = AdaBoostClassifier(n_estimators=15) # Low estimators to see the "cuts"
clf.fit(X, y)
```

```python
# 3. Create Meshgrid for Plotting
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.05),np.arange(y_min, y_max, 0.05))
```

```python
# 4. Predict on Meshgrid
Z = clf.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)
```

```python
# 5. Plot
plt.figure(figsize=(10, 6))
plt.contourf(xx, yy, Z, alpha=0.3, cmap='coolwarm')
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolors='k')
plt.title('Own Machine Learning Algorithm: AdaBoost (from scratch)\nDecision Boundary on Non-Linear Data')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```
